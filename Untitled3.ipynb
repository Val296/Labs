```python
import urllib.request
import os
from datetime import datetime

def download_vhi_data():
    # Создаем папку для данных, если её нет
    if not os.path.exists("vhi_data"):
        os.makedirs("vhi_data")

    # Список ID областей Украины (от 1 до 25 или 27 в зависимости от системы)
    for i in range(1, 28):
        # Проверяем, есть ли уже файл для этой области, чтобы не качать лишний раз
        existing_files = [f for f in os.listdir("vhi_data") if f.startswith(f"vhi_id_{i}_")]
        if existing_files:
            print(f"Файл для области {i} уже существует. Пропускаем.")
            continue

        # Ссылка на данные NOAA
        url = f"https://www.star.nesdis.noaa.gov/smcd/emb/vci/VH/get_provinceData.php?country=UKR&provinceID={i}&year1=1981&year2=2024&type=VHI"
        
        # Формируем имя файла с датой и временем
        now = datetime.now().strftime("%Y%m%d_%H%M%S")
        filename = f"vhi_id_{i}_{now}.csv"
        filepath = os.path.join("vhi_data", filename)

        try:
            urllib.request.urlretrieve(url, filepath)
            print(f"Область {i} успешно загружена: {filename}")
        except Exception as e:
            print(f"Ошибка при загрузке области {i}: {e}")

download_vhi_data()
```

```python
import pandas as pd
import glob
import os  # Обязательно добавьте этот импорт

def clean_data():
    # Ищем все CSV файлы в папке
    all_files = glob.glob("vhi_data/*.csv")
    list_of_dfs = ()

    for filename in all_files:
        # 1. Читаем файл, пропуская заголовок NOAA
        df = pd.read_csv(filename, index_col=None, header=1)
        
        # 2. Очищаем названия колонок от лишних пробелов
        df.columns = [c.strip() for c in df.columns]
        
        # 3. Удаляем строки с ошибками (где VHI = -1)
        df = df[df['VHI']!= -1]
        
        # 4. Получаем ID области из имени файла
        # Если файл называется "vhi_id_1_20260420.csv", то:
        # 'vhi'(0), 'id'(1), '1'(2), '20260420.csv'(3)
        # Поэтому используем индекс [1], а не [2]
        base_name = os.path.basename(filename)
        region_id = int(base_name.split('_')[1])
        
        # 5. Добавляем ID как новую колонку, а не заменяем весь df
        df['area_id'] = region_id
        
        list_of_dfs.append(df)

    # Соединяем все таблицы в одну
    if not list_of_dfs:
        print("Файлы не найдены!")
        return None
        
    return pd.concat(list_of_dfs, axis=0, ignore_index=True)

# Запуск и проверка
df_vhi = clean_data()
if df_vhi is not None:
    print("Первые 5 строк объединенной таблицы:")
    print(df_vhi.head())
```

```python
import pandas as pd
import glob
import os

def clean_data():
    # Ищем все файлы.csv в папке vhi_data
    all_files = glob.glob("vhi_data/*.csv")
    list_of_dfs = [] # ИСПРАВЛЕНО: создаем пустой список через

    for filename in all_files:
        # Читаем файл. 
        # sep=',' явно указываем разделитель-запятую
        # header=1 пропускает 1-ю строку (метаданные NOAA)
        # on_bad_lines='skip' заставит Pandas просто проигнорировать HTML-мусор в конце файла
        df = pd.read_csv(filename, index_col=None, header=1, sep=',', on_bad_lines='skip')
        
        # Очищаем названия колонок от лишних пробелов
        df.columns = [c.strip() for c in df.columns]
        
        # Удаляем строки, где VHI равен -1 (это ошибки в данных NOAA)
        # И убеждаемся, что в колонке VHI только числа
        df['VHI'] = pd.to_numeric(df['VHI'], errors='coerce')
        df = df[df['VHI']!= -1].dropna(subset=['VHI'])
        
        # Получаем ID области из имени файла
        # Если файл называется "vhi_id_1_2026...", то split('_') даст:
        # ['vhi', 'id', '1', '2026...']. Нам нужно число под индексом [3]
        base_name = os.path.basename(filename)
        try:
            # Пытаемся вытащить ID. Если индекс [3] не сработает, попробуйте [4]
            area_id = int(base_name.split('_')[3]) 
            df['area_id'] = area_id
            list_of_dfs.append(df)
        except Exception as e:
            print(f"Не удалось прочитать ID из файла {base_name}: {e}")

    if not list_of_dfs:
        print("Файлы не найдены или пусты! Проверьте папку vhi_data.")
        return None
        
    return pd.concat(list_of_dfs, axis=0, ignore_index=True)

# Запуск
df_vhi = clean_data()
if df_vhi is not None:
    print("Всё работает! Первые строки итоговой таблицы:")
    print(df_vhi.head())
```
